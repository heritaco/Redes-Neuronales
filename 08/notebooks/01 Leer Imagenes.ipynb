```python
import numpy as np
import os
import matplotlib.pyplot as plt
from PIL import Image    
```

```python
def Leer_Imagen(ruta, nombre, tamaño=None):
    """
    Carga una imagen desde un archivo y la convierte en un arreglo normalizado.

    Parámetros:
        ruta (str): Ruta del directorio donde se encuentra la imagen.
        nombre (str): Nombre del archivo de imagen (incluye la extensión).
        tamaño (tuple[int, int], opcional): Tamaño deseado (ancho, alto). 
                                            Si es None, mantiene el tamaño original.

    Retorna:
        np.ndarray: Imagen como arreglo NumPy normalizado en rango [0, 1].
    """
    img = Image.open(ruta + nombre)
    if tamaño is not None:
        img = img.resize(tamaño)
    imagen = np.asarray(img, dtype=np.float32) / 255.0
    return imagen

def Extraer_RGB(imagen):
    """
    Extrae los canales RGB de una imagen y los reorganiza en una matriz de píxeles.

    Parámetros:
        imagen (np.ndarray): Imagen en formato NumPy con 3 canales.

    Retorna:
        np.ndarray: Matriz de tamaño (N, 3), donde cada fila representa un píxel (R, G, B).
    """
    I1 = np.asarray(imagen, dtype=np.float32)
    R, G, B = I1[:, :, 0], I1[:, :, 1], I1[:, :, 2]
    XR = R.reshape((-1, 1))
    XG = G.reshape((-1, 1))
    XB = B.reshape((-1, 1))
    rgb = np.concatenate((XR, XG, XB), axis=1)
    return rgb

def Visualizar_Imagen(imagen):
    """
    Muestra una imagen sin ejes usando matplotlib.

    Parámetros:
        imagen (np.ndarray): Imagen en formato NumPy a mostrar.
    """
    plt.figure()
    plt.imshow(imagen)
    plt.axis('off')
    plt.title("Imagen original")
    plt.show()
    plt.close()
```

```python
def Listar_Archivos(ruta, extensiones=None):
    """
    Devuelve una lista con los nombres de todos los archivos dentro de una carpeta.
    Si se especifican extensiones (por ejemplo ['.jpg', '.png']), solo devuelve esos tipos.

    Parámetros:
        carpeta (str): Ruta de la carpeta a explorar.
        extensiones (list[str], opcional): Extensiones de archivo a filtrar.

    Retorna:
        list[str]: Lista con los nombres de los archivos encontrados.
    """
    archivos = []
    for nombre in os.listdir(ruta):
        ruta_completa = os.path.join(ruta, nombre)
        if os.path.isfile(ruta_completa):
            if extensiones is None or os.path.splitext(nombre)[1].lower() in extensiones:
                archivos.append(nombre)
    return archivos
```

```python
import cv2, numpy as np

def candidate_signature_rois(img, frac_h=0.22, frac_w=0.36):
    H, W = img.shape[:2]
    rois = []
    # esquinas inferiores
    for (y0,x0) in [(int((1-frac_h)*H), 0), (int((1-frac_h)*H), int((1-frac_w)*W)),
                    (int((1-frac_h)*H), int((1-frac_w/2)*W)),  # centro inferior
                    (int((1-frac_h)*H), int((1-frac_w* (1-0.5))*W))]:
        y1, x1 = min(H, y0+int(frac_h*H)), min(W, x0+int(frac_w*W))
        rois.append(img[y0:y1, x0:x1])
    return rois  # 3–4 candidatos

def signature_features(roi):
    g = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)
    g = cv2.GaussianBlur(g, (3,3), 0.8)
    edges = cv2.Canny(g, 60, 150)
    # cierre para unir trazos
    k = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
    edges = cv2.morphologyEx(edges, cv2.MORPH_CLOSE, k, iterations=1)
    # stats
    ink_ratio = edges.mean()  # densidad de trazo
    ys, xs = np.where(edges>0)
    orient_hist = np.zeros(12, dtype=np.float32)
    if len(xs)>0:
        gy, gx = np.gradient(g.astype(np.float32)/255.0)
        theta = np.rad2deg(np.arctan2(gy[ys, xs], gx[ys, xs])) % 180
        bins = (theta/15).astype(int)  # 12 bins
        for b in bins: orient_hist[b]+=1
        orient_hist /= orient_hist.sum()+1e-6
    # bordes del roi: grosor promedio de “stroke” por distancia a contorno
    dist = cv2.distanceTransform((255-edges).astype(np.uint8), cv2.DIST_L2, 3)
    if len(xs)>0: mean_stroke = dist[ys, xs].mean()
    else: mean_stroke = 0.0
    return np.concatenate([[ink_ratio, mean_stroke], orient_hist])
```

```python
# preprocess_art_pipeline.py
# Requisitos: pip install torch torchvision numpy opencv-python scikit-learn pillow
import argparse, os, sys, math, random, glob, io
from pathlib import Path
import numpy as np
import cv2
from PIL import Image
import torch
import torch.nn as nn
import torchvision.transforms as T
from torchvision.models import resnet50, ResNet50_Weights
from torchvision.models.feature_extraction import create_feature_extractor
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# -----------------------
# 1) Mapeos proporcionados
# -----------------------
painter_to_genre = {
    "Joan": "Surrealismo",
    "Salvador": "Surrealismo",
    "Rene": "Surrealismo",
    "Pablo": "Cubismo",
    "Jackson": "Expresionismo abstracto",
    "Rembrandt": "Barroco",
    "Caravaggio": "Barroco",
    "Camille": "Impresionismo",
    "Alfred": "Impresionismo",
    "Claude": "Impresionismo",
    "Vincent": "Impresionismo",
}

genre_to_label = {
    "Surrealismo": 0,
    "Cubismo": 1,
    "Expresionismo abstracto": 2,
    "Barroco": 3,
    "Impresionismo": 4,
}

# ---------------------------------------------
# 2) Utilidades: IO, color-constancy y recortes
# ---------------------------------------------
def list_images(root, exts=(".jpg", ".jpeg", ".png", ".bmp", ".webp")):
    root = Path(root)
    files = []
    for ext in exts:
        files.extend(map(str, root.rglob(f"*{ext}")))
    return sorted(files)

def read_rgb(path):
    # PIL maneja perfiles; devolvemos np.uint8 RGB
    img = Image.open(path).convert("RGB")
    return np.array(img)

def gamma_correct01(x01, gamma=2.2):
    # x01 float [0,1]
    x = np.clip(x01, 0, 1)
    return np.power(x, 1.0/gamma)

def shades_of_gray_cc(img_rgb, p=6, target_mean=0.5):
    # img_rgb uint8 -> float01
    I = img_rgb.astype(np.float32)/255.0
    I_p = np.power(I, p)
    rho = np.power(np.mean(I_p, axis=(0,1)) + 1e-8, 1.0/p)
    mu = target_mean
    I_cc = (mu / (rho + 1e-8)) * I
    I_cc = np.clip(I_cc, 0.0, 1.0)
    return (I_cc*255.0).astype(np.uint8)

def crop_for_embedding(img_rgb, pct=0.01):
    # Recorta ligeramente para embeddings, pero los stats de borde se extraen ANTES
    H, W = img_rgb.shape[:2]
    dy, dx = int(pct*H), int(pct*W)
    return img_rgb[dy:H-dy, dx:W-dx].copy()

# ------------------------------------
# 3) Características de firma y bordes
# ------------------------------------
def signature_candidate_rois(img_rgb, frac_h=0.22, frac_w=0.36):
    H, W = img_rgb.shape[:2]
    hh, ww = int(frac_h*H), int(frac_w*W)
    y0 = H - hh
    # esquinas y centro inferior
    xs = [0, max(0, W-ww), max(0, (W-ww)//2)]
    rois = []
    for x0 in xs:
        rois.append(img_rgb[y0:H, x0:x0+ww])
    return rois  # de 3 a 4

def signature_features(roi_rgb):
    # 14 dims: [ink_ratio, mean_stroke] + 12-bin hist de orientaciones
    g = cv2.cvtColor(roi_rgb, cv2.COLOR_RGB2GRAY)
    g = cv2.GaussianBlur(g, (3,3), 0.8)
    edges = cv2.Canny(g, 60, 150)
    k = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
    edges = cv2.morphologyEx(edges, cv2.MORPH_CLOSE, k, iterations=1)
    ink_ratio = edges.mean()  # [0,1] aprox
    ys, xs = np.where(edges>0)
    orient_hist = np.zeros(12, dtype=np.float32)
    if len(xs) > 0:
        gy, gx = np.gradient(g.astype(np.float32)/255.0)
        theta = (np.rad2deg(np.arctan2(gy[ys, xs], gx[ys, xs])) + 180.0) % 180.0
        bins = (theta // 15).astype(int)  # 12 bins de 15°
        for b in bins:
            orient_hist[b] += 1.0
        s = orient_hist.sum()
        if s > 0:
            orient_hist /= s
    dist = cv2.distanceTransform((255 - edges).astype(np.uint8), cv2.DIST_L2, 3)
    mean_stroke = float(dist[ys, xs].mean()) if len(xs)>0 else 0.0
    return np.concatenate([[ink_ratio, mean_stroke], orient_hist])

def aggregate_signature_features(img_rgb):
    rois = signature_candidate_rois(img_rgb)
    feats = [signature_features(r) for r in rois]
    F = np.stack(feats, axis=0) if len(feats)>0 else np.zeros((1,14), dtype=np.float32)
    # agregamos media y máximo → 28 dims
    return np.concatenate([F.mean(axis=0), F.max(axis=0)])

def border_frame_stats(img_rgb, pct=0.03):
    H, W = img_rgb.shape[:2]
    tH, tW = max(1,int(pct*H)), max(1,int(pct*W))
    top = img_rgb[:tH,:]; bottom = img_rgb[-tH:,:]
    left = img_rgb[:, :tW]; right = img_rgb[:, -tW:]
    def stats(b):
        m = b.mean(axis=(0,1))/255.0
        s = b.std(axis=(0,1))/255.0
        return np.concatenate([m, s])  # 6 dims por borde
    s = np.concatenate([stats(top), stats(bottom), stats(left), stats(right)])  # 24
    # textura de marco (proxy): promedio de std en gris
    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
    topg, botg, leftg, rightg = gray[:tH,:], gray[-tH:,:], gray[:, :tW], gray[:, -tW:]
    frame_texture = float(np.mean([topg.std(), botg.std(), leftg.std(), rightg.std()]) / 255.0)
    aspect = np.array([W/float(H), 1.0 if W>H else 0.0], dtype=np.float32)
    return np.concatenate([s, [frame_texture], aspect])  # 27 dims

# ------------------------------------
# 4) Color L*a*b* histogramas perceptual
# ------------------------------------
def lab_hist_features(img_rgb, bins=16):
    lab = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2LAB)
    Hs = []
    for c in range(3):
        h, _ = np.histogram(lab[...,c].ravel(), bins=bins, range=(0,255), density=True)
        Hs.append(h.astype(np.float32))
    return np.concatenate(Hs)  # 3*bins

# ------------------------------------
# 5) Embeddings CNN + Gram proyectado
# ------------------------------------
class StyleExtractor(nn.Module):
    """
    Extrae:
      - avgpool de ResNet50 (2048)
      - Gram pooling proyectado de layer1 y layer2 (32x32 cada uno -> 1024 + 1024)
    """
    def __init__(self, proj_out=32, seed=123):
        super().__init__()
        weights = ResNet50_Weights.IMAGENET1K_V2
        self.backbone = resnet50(weights=weights)
        self.backbone.eval()
        for p in self.backbone.parameters():
            p.requires_grad = False
        # puntos de extracción
        self.extractor = create_feature_extractor(
            self.backbone,
            return_nodes={
                "layer1": "l1",
                "layer2": "l2",
                "avgpool": "gap",
            }
        )
        # Proyecciones aleatorias fijas para reducir canales antes del Gram
        g = torch.Generator().manual_seed(seed)
        self.proj_out = proj_out
        self.R_l1 = nn.Parameter(torch.randn(256, proj_out, generator=g)*0.05, requires_grad=False)  # layer1 C=256
        self.R_l2 = nn.Parameter(torch.randn(512, proj_out, generator=g)*0.05, requires_grad=False)  # layer2 C=512

        # Normalización ImageNet
        self.tf = T.Compose([
            T.Resize(560, interpolation=T.InterpolationMode.BICUBIC),
            T.CenterCrop(448),
            T.ToTensor(),
            T.Normalize(mean=weights.meta["mean"], std=weights.meta["std"]),
        ])

    @staticmethod
    def gram_from_feature(F):  # F: (C', H, W)
        C, H, W = F.shape
        X = F.view(C, -1)                 # (C, HW)
        G = (X @ X.t()) / (H*W)           # (C, C)
        g = G.reshape(-1)
        return g

    def forward(self, pil_image):
        # Espera PIL.Image RGB
        x = self.tf(pil_image).unsqueeze(0)
        with torch.no_grad():
            feats = self.extractor(x)
            gap = feats["gap"].view(1, -1).squeeze(0)  # (2048,)
            l1 = feats["l1"].squeeze(0)               # (256,H1,W1)
            l2 = feats["l2"].squeeze(0)               # (512,H2,W2)
            # Proyección de canales
            # (C,H,W) -> (C',H,W) con proyección lineal sobre C
            l1p = torch.einsum("chw,cp->phw", l1, self.R_l1)  # (proj_out,H1,W1)
            l2p = torch.einsum("chw,cp->phw", l2, self.R_l2)  # (proj_out,H2,W2)
            g1 = self.gram_from_feature(l1p)  # (proj_out^2,)
            g2 = self.gram_from_feature(l2p)  # (proj_out^2,)
            out = torch.cat([gap, g1, g2], dim=0)  # 2048 + 1024 + 1024 = 4096
        return out.cpu().numpy().astype(np.float32)

# ------------------------------------
# 6) Ensamble de características por imagen
# ------------------------------------
def features_for_image(path, style_extractor, do_cc=True, gamma=2.2, crop_pct=0.01):
    # Lectura
    img_rgb = read_rgb(path)
    H, W = img_rgb.shape[:2]
    # 6.1 Stats de marco/bordes + aspecto ANTES de recortar
    feat_border = border_frame_stats(img_rgb, pct=0.03)  # 27
    # 6.2 Firma
    feat_sign = aggregate_signature_features(img_rgb)    # 28
    # 6.3 Normalización fotométrica para el embedding y Lab
    work = img_rgb.copy()
    work = gamma_correct01(work.astype(np.float32)/255.0, gamma=gamma)
    work = (work*255.0).astype(np.uint8)
    if do_cc:
        work = shades_of_gray_cc(work, p=6, target_mean=0.5)
    # 6.4 Recorte leve para el embedding
    emb_img = crop_for_embedding(work, pct=crop_pct)
    # 6.5 Color perceptual
    feat_lab = lab_hist_features(emb_img, bins=16)       # 48
    # 6.6 Embeddings estilo (avgpool + Gram proyectado)
    pil_img = Image.fromarray(emb_img)
    feat_style = style_extractor(pil_img)                # 4096
    # 6.7 Concatenación
    ratio = np.array([W/float(H)], dtype=np.float32)     # 1
    z = np.concatenate([feat_style, feat_lab, feat_sign, feat_border, ratio], axis=0)
    return z  # dim = 4096 + 48 + 28 + 27 + 1 = 4200

# ------------------------------------
# 7) Etiquetado desde el nombre de archivo
# ------------------------------------
def infer_label_from_filename(path):
    name = Path(path).stem
    # painter = token inicial antes de '_'
    painter = name.split("_")[0]
    # tolerante: si no está, intenta encontrar clave contenida
    if painter not in painter_to_genre:
        for k in painter_to_genre.keys():
            if name.startswith(k):
                painter = k
                break
    genre = painter_to_genre.get(painter, None)
    if genre is None:
        raise ValueError(f"No se pudo mapear pintor en: {path}")
    label = genre_to_label[genre]
    return label

# ------------------------------------
# 8) Pipeline: extraer, estandarizar, PCA-whiten
# ------------------------------------
def build_dataset(data_dir, pca_dim=256, seed=123, cc=True, gamma=2.2, crop_pct=0.01):
    rng = np.random.RandomState(seed)
    torch.manual_seed(seed)
    random.seed(seed)

    files = list_images(data_dir)
    if len(files) == 0:
        raise RuntimeError(f"No se encontraron imágenes en {data_dir}")

    device = "cuda" if torch.cuda.is_available() else "cpu"
    style_extractor = StyleExtractor(proj_out=32, seed=seed).to("cpu")  # CPU suficiente para N~200

    feats = []
    labels = []
    for p in files:
        try:
            z = features_for_image(p, style_extractor, do_cc=cc, gamma=gamma, crop_pct=crop_pct)
            y = infer_label_from_filename(p)
            feats.append(z); labels.append(y)
        except Exception as e:
            print(f"[WARN] salto {p}: {e}")

    X = np.stack(feats, axis=0).astype(np.float32)   # (N, 4200)
    y = np.array(labels, dtype=np.int64)             # (N,)

    # Estandarización + PCA (whiten)
    scaler = StandardScaler(with_mean=True, with_std=True)
    Xs = scaler.fit_transform(X)
    pca = PCA(n_components=pca_dim, whiten=True, random_state=seed)
    Xp = pca.fit_transform(Xs).astype(np.float32)    # (N, pca_dim)

    # Pesos de clase sugeridos: N/(K*n_k)
    K = len(genre_to_label)
    counts = np.bincount(y, minlength=K).astype(np.float32)
    N = float(len(y))
    class_weights = (N / (K * np.maximum(counts, 1.0))).astype(np.float32)

    meta = {
        "files": files,
        "counts_per_class": counts,
        "class_weights": class_weights,
        "scaler_mean": scaler.mean_,
        "scaler_scale": scaler.scale_,
        "pca_components": pca.components_.astype(np.float32),
        "pca_mean": pca.mean_.astype(np.float32),
        "pca_explained_variance": pca.explained_variance_.astype(np.float32),
        "label_map": genre_to_label,
    }
    return Xp, y, meta

# ------------------------------------
# 9) CLI
# ------------------------------------
def main():
    ap = argparse.ArgumentParser("Preprocesamiento pro-arte para géneros")
    ap.add_argument("--data_dir", type=str, required=True, help="Carpeta con imágenes")
    ap.add_argument("--out", type=str, default="features_art.npz", help="Ruta de salida .npz")
    ap.add_argument("--pca_dim", type=int, default=256, help="Dimensión final PCA-whiten")
    ap.add_argument("--no_cc", action="store_true", help="Desactiva constancia de color")
    ap.add_argument("--gamma", type=float, default=2.2, help="Gamma para corrección")
    ap.add_argument("--crop_pct", type=float, default=0.01, help="Recorte leve para embeddings")
    ap.add_argument("--seed", type=int, default=123, help="Semilla")
    args = ap.parse_args()

    X, y, meta = build_dataset(
        args.data_dir, pca_dim=args.pca_dim, seed=args.seed,
        cc=(not args.no_cc), gamma=args.gamma, crop_pct=args.crop_pct
    )

    np.savez_compressed(
        args.out,
        X=X, y=y,
        files=np.array(meta["files"]),
        counts_per_class=meta["counts_per_class"],
        class_weights=meta["class_weights"],
        scaler_mean=meta["scaler_mean"], scaler_scale=meta["scaler_scale"],
        pca_components=meta["pca_components"], pca_mean=meta["pca_mean"],
        pca_explained_variance=meta["pca_explained_variance"],
        label_map=np.array(list(genre_to_label.items()), dtype=object),
    )
    print(f"OK: X.shape={X.shape}, y.shape={y.shape}")
    print("counts_per_class:", meta["counts_per_class"])
    print("class_weights (N/(K*n_k)):", meta["class_weights"])
    print(f"Guardado en: {args.out}")

if __name__ == "__main__":
    main()
```
