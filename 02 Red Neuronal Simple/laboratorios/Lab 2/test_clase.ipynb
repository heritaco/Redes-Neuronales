## Datos 

Los problemas que se van a resolver son los siguientes: AND, XOR y un set de datos provistos por el profesor.


```python
# Datos de entrada 
AND = [[1, 1],
     [0, 1],
     [1, 0],
     [0, 0]]

T_and= [1, -1, -1, -1]  # Salidas esperadas
```

```python
# Datos de entrada 
XOR = [[1, 1],
     [0, 1],
     [1, 0],
     [0, 0]]

T_xor= [1, -1, -1, 1]  # Salidas esperadas
```

# Perceptron Simple

En esta sección se va a probar la clase `Perceptron` que fue implementada en el script perceptron.py. Se muestran los entrenamientos, ajuste y animacion por iteraciones de los datos AND y XOR.


```python
# Crear y entrenar el perceptrón
perceptron = Perceptron(AND,T= T_and, learning_rate=0.1, epochs=30, umbral=0.01, weights=[0.5, 0.5], bias=0.5)
perceptron.fit()
perceptron.plot()
```

```python
# Crear y entrenar el perceptrón
perceptron = Perceptron(XOR, T =T_xor, learning_rate=0.1, epochs=30, umbral=0.01, weights=[1, 1], bias=-0.5)
perceptron.fit()
perceptron.plot()
```

Además se muestran los métodos y atributos que se pueden llamar de la clase `Perceptron`.

```python
perceptron.get_bias()
```

```python
perceptron.get_weights()
```

## ADALINE

En esta sección se va a probar la clase `Adaline` que fue implementada en el script adaline.py. Se muestran los entrenamientos, ajuste y animacion por iteraciones de los datos AND y XOR.

```python
adaline = Adaline(AND, T = T_and, learning_rate=0.1, epochs=1, umbral=0.1, tolerance=1e-36,weights=[1, 1], bias=-0.5)
adaline.fit()
adaline.plot()
```

```python
adaline = Adaline(XOR, T = T_xor, learning_rate=0.1, epochs=1, umbral=0.1, tolerance=1e-36,weights=[1, 1], bias=-0.5)
adaline.fit()
adaline.plot()
```

Adaline tiene la posibilidad de llamar a los mismos metodos y atributos que Perceptron.

```python
adaline.get_weights()
```

```python
adaline_malos = adaline.get_malos()
adaline_malos[0]  # Datos mal clasificados en la primera iteración
```

```python
len(adaline_malos[0])  # número de datos mal clasificados en la primera iteración
```

## Prueba con datos Cáncer de Sklearn.

Crea una tabla donde muestres el porcentaje de instancias clasificadas correctamente para 5 valores diferentes de tasas de aprendizaje y 3 valores de umbral.

```python
data = load_breast_cancer()
df_breast_cancer = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
pd.set_option('display.max_columns', None)
display(df_breast_cancer)
df_breast_cancer['target'] = data.target
df_breast_cancer.head()
```

```python
S = df_breast_cancer[['mean radius', 'mean texture']].values  # array de vectores de 2 entradas

# El vector de los targets
T = df_breast_cancer['target'].values
```

```python
Perceptron = Perceptron(S, T, learning_rate=0.1, epochs=1, umbral=0.1, weights=[.75, 1], bias=-0.5)
Perceptron.fit()
Perceptron.plot()
```

```python
Adaline = Adaline(S, T, learning_rate=0.1, epochs=1, umbral=0.1, tolerance=1e-36,weights=[.75, 1], bias=-0.5)
Adaline.fit()
Adaline.plot()
```
