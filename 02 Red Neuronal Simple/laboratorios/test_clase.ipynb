```python
from perceptron import Perceptron
from adaline import ADALINE
```

```python
# Datos de entrada 
S = [[1, 3],
     [3, 3],
     [1, 1],
     [3, 1],
     [-2, -1],
     [-1, -2]]

T = [1, -1, -1, -1, 1, 1]  # Salidas esperadas
```

```python
# Crear y entrenar el perceptrón
perceptron = Perceptron(S, T, learning_rate=0.05, epochs=10, umbral=0.5)
perceptron.fit()
# perceptron.plot()
```

```python
perceptron.get_bias()
```

```python
perceptron.get_weights()
```

```python
malos = perceptron.get_malos()  # Número de clasificaciones incorrectas
```

```python
malos[0]
```

```python
adaline = ADALINE(S, T, learning_rate=0.0001, epochs=10, umbral=0.1, tolerance=1e-16)
adaline.fit()
adaline.plot()
```
