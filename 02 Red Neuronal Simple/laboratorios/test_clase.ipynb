```python
from perceptron import Perceptron
from adaline import ADALINE
```

```python
import style
style.apply()
```

```python
# Datos de entrada 
S = [[1, 3],
     [3, 3],
     [1, 1],
     [3, 1],
     [-2, -1],
     [-1, -2]]

T = [1, -1, -1, -1, 1, 1]  # Salidas esperadas
```

```python
# Crear y entrenar el perceptrón
perceptron = Perceptron(S, T, learning_rate=0.05, epochs=10, umbral=0.5)
perceptron.fit()
# perceptron.plot()
```

```python
perceptron.get_bias()
```

```python
perceptron.get_weights()
```

```python
malos = perceptron.get_malos()  # Número de clasificaciones incorrectas
```

```python
malos[6]
```

```python
len(malos[6]) #  número de datos mal clasificados en cada iteración
```

```python
perceptron.plot()
```

```python
adaline = ADALINE(S, T, learning_rate=0.0001, epochs=1, umbral=0.1, tolerance=1e-16)
adaline.fit()
#adaline.plot()
```

```python
adaline.get_bias()
```

```python
adaline.get_weights()
```

```python
adaline_malos = adaline.get_malos()
adaline_malos[0]  # Datos mal clasificados en la primera iteración
```
