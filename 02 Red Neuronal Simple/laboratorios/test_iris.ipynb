```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
from perceptron import Perceptron
from Adaline import adaline
import style 
style.mpl_apply()
```

```python
iris = datasets.load_iris()                                     # load the iris dataset
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)   # convert to a pandas dataframe
df['target'] = iris.target                                      # add the target column
df = df.iloc[0:100, [0, 1, -1]]                                 # select only the first 100 rows and the first two columns
data = df.to_numpy()                                            # convert to numpy array

S = data[:, 0:2]                                                # split the data into features and target
T = data[:, 2]
T = np.where(T == 0, -1, 1)                                     # convert the target to -1 and 1
```

```python
plt.figure()
plt.scatter(S[:50, 0], S[:50, 1], marker='o', label='Iris-setosa')  # plot the data
plt.scatter(S[50:100, 0], S[50:100, 1], marker='x', label='Iris-versicolor')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.legend()
plt.title('Iris Dataset - First 100 samples')
plt.show()
```

```python
print("Shape of S:", S.shape)                                   # print the shape of the data
print("Shape of T:", T.shape)
print("First 5 rows of S:\n", S[0:5, :])                        # print the first 5 rows of the data
print("First 5 rows of T:\n", T[0:5])
```

```python
# create the perceptron object
perceptron = Perceptron(S, T, learning_rate=0.8, epochs=3, umbral=0.5, weights=np.array([1, 2]), bias=0.3)
# train the perceptron
perceptron.fit()
# plot the decision boundary
perceptron.plot()
```
